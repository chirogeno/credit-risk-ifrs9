# Carga de datos de Lending Club
Cargo el CSV en DuckDB y compruebo que los datos son préstamos válidos.

```python
import duckdb

con = duckdb.connect("../data/lending_club.duckdb")
```

```python
con.sql("SELECT COUNT(*) FROM (DESCRIBE prestamos)")
```

```python
con.sql("""
    SELECT COUNT(*)
    FROM prestamos
    WHERE loan_amnt IS NULL
      AND id LIKE 'Total amount%'
""")
```

```python
con.sql("""
    SELECT id, loan_amnt
    FROM prestamos
    WHERE loan_amnt IS NULL
      AND (id IS NULL OR id NOT LIKE 'Total amount%')
""")
```

```python
con.sql("""
    CREATE OR REPLACE TABLE prestamos AS
    SELECT *
    FROM prestamos
    WHERE loan_amnt IS NOT NULL
""")
```
